```python
def hammingDistance(p: str, q: str) -> int:
    count = 0

    # This is based on the assumption that p and q have the same length
    for i in range(len(p)):
        if p[i] != q[i]:
            count += 1

    return count

with open("rosalind_ba1g.txt", "r") as file:
	str_list = []
	for line in file:
		line = line.rstrip()
		str_list.append(line)

	p = str_list[0]
	q = str_list[1]

print(hammingDistance(p, q))
```

```python
def approximatePatternMatching(pattern: str, text: str, d: int) -> list[int]:
    count = 0

    for i in range(len(text)):
        if hummingDistance(pattern, text[i:i+len(pattern)]) <= d:
            count += 1

    return count


with open("rosalind_ba1h.txt", "r") as file:
	str_list = []
	for line in file:
		line = line.rstrip()
		str_list.append(line)

	pattern = str_list[0]
	text = str_list[1]
    d = int(str_list[2])

print(approximatePatternMatching(pattern, text, d))
```
