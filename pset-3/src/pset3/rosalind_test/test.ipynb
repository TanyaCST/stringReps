## `find_clumps`

```python
def findClumps(genome: str, k: int, L: int, t: int) -> str:
    """Return distinct k-mers appearing at least t times in any length-L window.

    Count overlapping occurrences fully contained within each window.
    Assume uppercase DNA, 1 <= k <= L <= len(genome), and t >= 1.
    Each qualifying k-mer appears only once in the returned set.
    """
    kmer_table = {}

    # Set up a frequency dictionary for each k mer existing in the range
    for i in range(L):
        if genome[i:i+k] not in kmer_table:
            kmer_table[genome[i:i+k]] = 1
        else:
            kmer_table[genome[i:i+k]] += 1

    output = ""

    # Find the list of at least t times frequent k mer
    for kmer in kmer_table:
        if kmer_table[kmer] >= t:
            output += kmer
            output += " "

    return output
```

```python
with open("rosalind_ba1e.txt", "r") as file:
	str_list = []
	for line in file:
		line = line.rstrip()
		str_list.append(line)

	text = str_list[0]
	numbers = str_list[1]
	k, L, t = numbers.split()

print(type(k), type(L), type(t))

print(findClumps(text, int(k), int(L), int(t)))
```

## `frequency_array`

```python
def frequencyArray(text: str, k: int) -> list[int]:
    """Return counts for all 4**k DNA k-mers in lexicographic order.

    Use alphabet order A, C, G, T. Include zero counts and count
    overlapping occurrences. Assume uppercase DNA and 1 <= k <= len(text).
    """
    # Pseudocode
    # 1. Create a frequency table of all kmers
    # 2. Sort the keys of the table with merge sort (actually I will just use python sort here)
    #    --> This I think will be feasible since each str is a number and the size is ascending as alphabetical order
    # 3. Create a empty frequency arraylist
    #    --> Iterate through the sorted keys and append them into the arraylist

    # Set up a frequency table for each k mer existing in the range
    kmer_table = {}

    for i in range(len(text)):
        if text[i:i+k] not in kmer_table:
            kmer_table[text[i:i+k]] = 1
        else:
            kmer_table[text[i:i+k]] += 1

    sorted_kmer = sorted(kmer_table)

    frequency_arr = []

    for kmer in sorted_kmer:
        frequency_arr.append(kmer_table[kmer])

    return frequency_arr
```

```python
with open("rosalind_ba1k.txt", "r") as file:
	str_list = []
	for line in file:
		line = line.rstrip()
		str_list.append(line)

	text = str_list[0]
	k = int(str_list[1])

print(frequencyArray(text, k))
```

## `reverse_complement`

```python
def reverseComplement(text: str) -> str:
    """Return the reverse complement of an uppercase DNA string.

    A pairs with T, and C pairs with G. Return the complementary
    nucleotides in reverse order.
    """

    # From the tail of text, append its reverse

    reverse = ""

    i = len(text) - 1

    while i >= 0:
        if text[i] == "A":
            reverse += "T"

        if text[i] == "T":
            reverse += "A"

        if text[i] == "C":
            reverse += "G"

        if text[i] == "G":
            reverse += "C"

        i -= 1

    return reverse
```

```python
with open("rosalind_ba1c.txt", "r") as file:
	str_list = []
	for line in file:
		line = line.rstrip()
		str_list.append(line)

	text = str_list[0]

print(reverseComplement(text))
```

## `transcribe_rna`

```python
def transcribeRNA(text: str) -> str:
    """Return the RNA string corresponding to an uppercase DNA coding strand.

    Replace each T with U, retaining the order of the other nucleotides.
    """
    rna = ""
    for i in range(len(text)):
        if text[i] == "T":
            rna += "U"
        else:
            rna += text[i]

    return rna
```

```python
with open("rosalind_rna.txt", "r") as file:
	str_list = []
	for line in file:
		line = line.rstrip()
		str_list.append(line)

	text = str_list[0]

print(transcribeRNA(text))
```
